```python
df_dropped = df.dropna()

display(df_dropped)
```

```python
df_filled = df.copy()

df_filled["area"] = df_filled["area"].fillna(
    df_filled["area"].mean()
)

display(df_filled)
```

```python
df_t = df.copy()

# apply(): create new column
df_t["price_per_sqft"] = df_t.apply(
    lambda row: row["price"] / row["area"],
    axis=1
)

# map(): convert yes/no into 1/0
yes_no_map = {
    "yes": 1,
    "no": 0
}

df_t["mainroad_code"] = df_t["mainroad"].map(yes_no_map)

# astype(): change datatype
df_t["bedrooms"] = df_t["bedrooms"].astype(int)

# rename(): rename column
df_t = df_t.rename(
    columns={"price": "house_price"}
)

# sort_values(): highest price first
df_t = df_t.sort_values(
    by="house_price",
    ascending=False
)

df_t
```

```python
furnishing_info = pd.DataFrame({
    "furnishingstatus": [
        "furnished",
        "semi-furnished",
        "unfurnished"
    ],
    "furnishing_code": [3, 2, 1]
})

furnishing_info
```

```python
merged_df = pd.merge(
    df,
    furnishing_info,
    on="furnishingstatus",
    how="left"
)

print("merge() result:")

display(merged_df.head())
```

```python
joined_df = df.set_index("furnishingstatus").join(
    furnishing_info.set_index("furnishingstatus"),
    how="left"
)

print("join() result:")

display(joined_df.head())
```
